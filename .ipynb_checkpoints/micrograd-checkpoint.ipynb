```python
import math
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
def f(x):
    return 3*np.sin(x) + 4*x**2 + 7*x - 2 
def df(x):
    return 3*np.cos(x) + 8*x + 7
```

```python
xs = np.arange(-6,6, 0.25)
ys = f(xs)
plt.plot(xs,ys)
dys1 = df(xs)
```

```python
dys = np.zeros(len(xs))
for i in range(len(ys)-1):
    diff = ys[i+1]-ys[i]
    dys[i] = diff/0.25
plt.plot(xs,dys)
```

```python
plt.plot(xs,dys1)
```

```python
#for multivariate
def f2(a,b,c):
    return a*b + c

h = 0.0001
a = 3
b = -4
c = 2.4

#wrt a
d_a = (f2(a+h,b,c)-f2(a,b,c))/h
#similarly
d_b = (f2(a,b+h,c)-f2(a,b,c))/h
d_c = (f2(a,b,c+h)-f2(a,b,c))/h

print(f"{d_a} {d_b} {d_c}")
```

```python
class Value:
    def __init__(self, data):
        self.data = data
    def __repr__(self):
        return f"Value(data:{self.data})"
    def __add__(self,other):
        return Value(self.data+other.data)
    def __mul__(self,other):
        return Value(self.data*other.data)

h = 0.0001
a = 3
b = -4
c = 2.4

print(Value(a)+Value(b))
```
